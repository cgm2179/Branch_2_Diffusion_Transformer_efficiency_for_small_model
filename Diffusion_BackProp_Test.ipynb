Imports

```python
import torch
import matplotlib.pyplot as plt
import deepinv as dinv
import os
from torchvision import datasets, transforms
```

Device

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

print(device)
```

Recreate the diffusion schedule

```python
timesteps = 1000

beta_start = 1e-4
beta_end = 0.02

betas = torch.linspace(beta_start, beta_end, timesteps).to(device)

alphas = 1.0 - betas

alphas_cumprod = torch.cumprod(alphas, dim=0)

sqrt_alphas_cumprod = torch.sqrt(alphas_cumprod)

sqrt_one_minus_alphas_cumprod = torch.sqrt(1 - alphas_cumprod)
```

Load the test dataset

```python
transform = transforms.Compose([
    transforms.Resize(32),
    transforms.ToTensor(),
    transforms.Normalize((0.0,), (1.0,))
])

test_loader = torch.utils.data.DataLoader(
    datasets.MNIST(
        "./data",
        train=False,
        download=True,
        transform=transform
    ),
    batch_size=8,
    shuffle=True
)
```

oad your trained model

```python

model = dinv.models.DiffUNet(
    in_channels=1,
    out_channels=1,
    pretrained=None
).to(device)

model.load_state_dict(
    torch.load("/content/trained_diffusion_model.pth", map_location=device)
)

model.eval()

print("Model Loaded")
```

### Correctly Saving and Loading a PyTorch Model

When saving a PyTorch model, it's generally recommended to save only the `state_dict` (the learned parameters) rather than the entire model object. This approach offers more flexibility, especially if the model architecture changes. When loading, you'll first define the model architecture and then load the saved `state_dict` into it.

```python
# First, let's create a dummy model for demonstration purposes
class DummyModel(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = torch.nn.Linear(10, 1)

    def forward(self, x):
        return self.linear(x)

dummy_model = DummyModel().to(device)

# Define a path to save the model
MODEL_PATH = '/content/dummy_model.pth'

# Save only the model's state_dict
torch.save(dummy_model.state_dict(), MODEL_PATH)

print(f"Dummy model state_dict saved to {MODEL_PATH}")
```

```python
# Now, let's load the model back

# 1. Re-instantiate the model with the same architecture
loaded_model = DummyModel().to(device)

# 2. Load the saved state_dict into the new model instance
loaded_model.load_state_dict(torch.load(MODEL_PATH, map_location=device))

# 3. Set the model to evaluation mode (important for inference)
loaded_model.eval()

print(f"Model loaded successfully from {MODEL_PATH}")
print("Loaded model state_dict keys:", loaded_model.state_dict().keys())
```

This method ensures that your model checkpoint is saved and loaded in a robust way, reducing the chances of corruption.

Get a batch of test images

```python
images, labels = next(iter(test_loader))

images = images.to(device)

print(images.shape)
```

Add random Gaussian noise

```python
noise = torch.randn_like(images)

t = torch.randint(
    0,
    timesteps,
    (images.size(0),),
    device=device
)

noisy_images = (
    sqrt_alphas_cumprod[t,None,None,None] * images +
    sqrt_one_minus_alphas_cumprod[t,None,None,None] * noise
)
```

Predict the noise

```python
with torch.no_grad():

    predicted_noise = model(
        noisy_images,
        t,
        type_t="timestep"
    )
```

Compute the test loss

```python
criterion = torch.nn.MSELoss()

loss = criterion(predicted_noise, noise)

print("Noise Prediction Loss:", loss.item())
```

Cell 10 – Recover the original image

The DDPM forward process is

xt​=αˉtx0+1−αˉtϵ.

Rearranging gives

x^0=αˉtxt−1−αˉtϵ^

```python
recovered = (
    noisy_images -
    sqrt_one_minus_alphas_cumprod[t,None,None,None]
    * predicted_noise
) / sqrt_alphas_cumprod[t,None,None,None]
```

Display the results

```python
fig, axes = plt.subplots(3,8, figsize=(16,6))

for i in range(8):

    axes[0,i].imshow(
        images[i,0].cpu(),
        cmap="gray"
    )
    axes[0,i].set_title("Original")
    axes[0,i].axis("off")

    axes[1,i].imshow(
        noisy_images[i,0].cpu(),
        cmap="gray"
    )
    axes[1,i].set_title("Noisy")
    axes[1,i].axis("off")

    axes[2,i].imshow(
        recovered[i,0].cpu(),
        cmap="gray"
    )
    axes[2,i].set_title("Recovered")
    axes[2,i].axis("off")

plt.tight_layout()
plt.show()
```

Compare the true and predicted noise

```python
fig, axes = plt.subplots(2,8, figsize=(16,4))

for i in range(8):

    axes[0,i].imshow(
        noise[i,0].cpu(),
        cmap="gray"
    )
    axes[0,i].set_title("True")
    axes[0,i].axis("off")

    axes[1,i].imshow(
        predicted_noise[i,0].cpu(),
        cmap="gray"
    )
    axes[1,i].set_title("Pred")
    axes[1,i].axis("off")

plt.tight_layout()
plt.show()
```
